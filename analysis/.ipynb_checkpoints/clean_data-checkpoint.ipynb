```python
import pandas as pd

d = pd.read_csv("../data/raw_sheet_2026-10-26.csv")
d.head()
```

```python
print("Rows:", len(d))
print("A values:", sorted(d.A.unique()))
print("Participants:", d.participant.nunique())
```

```python
d.loc[d.A == 1, "A"] = 250
d.loc[d.participant == "P5518", "participant"] = "P3784"
```

```python
print("Rows:", len(d))
print("A values:", sorted(d.A.unique()))
print("Participants:", d.participant.nunique())
```

```python
fast_miss = (d.hit == 0) & (d.MT_ms < 250)
print("Fast misses:", fast_miss.sum())
d[fast_miss][["participant", "device", "A", "W", "MT_ms", "clickX", "targetX"]]
```

```python
old_targetX = 1200 - d.targetX
closer_to_old = (d.clickX - old_targetX).abs() < (d.clickX - d.targetX).abs()

d["double_tap"] = fast_miss & closer_to_old
print("Double taps:", d.double_tap.sum())
```

```python
d.groupby(["participant", "device"]).double_tap.sum()
```

```python
clean = d[~d.double_tap].copy()
print("Rows before:", len(d))
print("Rows after: ", len(clean))
```

```python
hits = clean[clean.hit == 1].sort_values(["run", "trial"]).copy()
hits["hit_no"] = hits.groupby(["run", "A", "W"]).cumcount() + 1
print("Hits:", len(hits))

hits = hits[hits.hit_no > 1]
print("Hits without first movement:", len(hits))
```

```python
hits["card_median"] = hits.groupby(["run", "A", "W"]).MT_ms.transform("median")
hits["outlier"] = hits.MT_ms > 3 * hits.card_median
print("Outliers:", hits.outlier.sum())

hits = hits[~hits.outlier]
print("Hits left:", len(hits))
```

```python
clean.to_csv("../data/clean_clicks.csv", index=False)
hits.to_csv("../data/clean_hits.csv", index=False)
print("Saved.")
```
